```python
import re
```

```python
def extract_skills(description):
    if pd.isna(description):
        return []

    text = description.lower()

    found_skills = []

    for category, skills in skills_taxonomy.items():
        for skill in skills:
            skill_lower = skill.lower()

            # Escape special characters such as +, #, .
            pattern = r'(?<!\w)' + re.escape(skill_lower) + r'(?!\w)'

            if re.search(pattern, text):
                found_skills.append(skill)

    return list(set(found_skills))
```

```python
def normalize_skills(found_skills):
    normalized = []

    for skill in found_skills:
        key = skill.lower()

        if key in skill_normalization:
            normalized.append(skill_normalization[key])
        else:
            normalized.append(skill)

    return list(set(normalized))
```

```python
def classify_role(title):

    title = str(title).lower()

    if any(k in title for k in [
        'data analyst',
        'data analytics',
        'business analyst'
    ]):
        return 'Data Analytics'

    elif any(k in title for k in [
        'machine learning',
        'ai engineer',
        'data scientist',
        'ml engineer'
    ]):
        return 'AI/ML'

    elif any(k in title for k in [
        'software engineer',
        'software developer',
        'sde',
        'full stack',
        'backend',
        'frontend'
    ]):
        return 'Software Development'

    else:
        return 'Other'


target_titles_clean['role_category'] = (
    target_titles_clean['title'].apply(classify_role)
)

print(target_titles_clean['role_category'].value_counts())
```

```python
import pandas as pd

job_skills_gt = pd.read_csv("../data/raw/job_skills.csv")
skills_gt = pd.read_csv("../data/raw/skills.csv")

print("job_skills shape:", job_skills_gt.shape)
print("skills shape:", skills_gt.shape)

print("\njob_skills columns:")
print(job_skills_gt.columns.tolist())

print("\nskills columns:")
print(skills_gt.columns.tolist())

print("\nSkills sample:")
print(skills_gt.head())
```

```python
# Convert skill abbreviations into actual skill names
ground_truth = job_skills_gt.merge(
    skills_gt,
    on='skill_abr',
    how='left'
)

print(ground_truth.head())
print("\nGround-truth rows:", len(ground_truth))
print("Unique jobs:", ground_truth['job_id'].nunique())
print("Unique skills:", ground_truth['skill_name'].nunique())
```

```python
print(
    "Missing skill-name mappings:",
    ground_truth['skill_name'].isna().sum()
)
```

```python
# Get 25 job IDs from our analyzed postings
sample_job_ids = target_titles_clean['job_id'].dropna().sample(
    25,
    random_state=42
)

# Ground-truth skills for those jobs
gt_sample = ground_truth[
    ground_truth['job_id'].isin(sample_job_ids)
]

# Our extracted skills from MySQL
our_sample = pd.read_sql("""
    SELECT job_id, skill_name
    FROM posting_skills
""", engine)

our_sample = our_sample[
    our_sample['job_id'].isin(sample_job_ids)
]

print("Ground-truth rows:", len(gt_sample))
print("Our extracted rows:", len(our_sample))
```

```python
# Convert each job's skills into sets
gt_dict = (
    gt_sample.groupby('job_id')['skill_name']
    .apply(set)
    .to_dict()
)

our_dict = (
    our_sample.groupby('job_id')['skill_name']
    .apply(set)
    .to_dict()
)

results = []

for job_id in sample_job_ids:

    gt_skills = gt_dict.get(job_id, set())
    our_skills = our_dict.get(job_id, set())

    matched = gt_skills & our_skills
    missing = gt_skills - our_skills
    extra = our_skills - gt_skills

    results.append({
        'job_id': job_id,
        'ground_truth_count': len(gt_skills),
        'our_count': len(our_skills),
        'matched': len(matched),
        'missing': len(missing),
        'extra': len(extra)
    })

validation = pd.DataFrame(results)

print(validation)
```

```python
def classify_role(title):
    title = str(title).lower()

    if any(x in title for x in [
        "machine learning", "ml engineer", "data scientist",
        "ai engineer", "artificial intelligence"
    ]):
        return "AI/ML"

    elif any(x in title for x in [
        "data analyst", "business analyst", "data analytics"
    ]):
        return "Data Analytics"

    elif any(x in title for x in [
        "software engineer", "software developer"
    ]):
        return "Software Development"

    return "Other"


postings["role_category"] = postings["title"].apply(classify_role)

print(postings["role_category"].value_counts())
```
